```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("dataset_2190_cholesterol.csv")
print(df.head())
```

```python
first_col = df.columns[0]
last_col = df.columns[-1]

x = df[first_col]
y = df[last_col]
```

```python
plt.plot(x, y)
plt.title("Line Plot")
plt.xlabel(first_col)
plt.ylabel(last_col)
plt.savefig("line_plot.png")
plt.close()
```

```python
x.value_counts().plot(kind='bar')
plt.title("Bar Chart of First Column")
plt.savefig("bar_chart.png")
plt.close()
```

```python
plt.hist(y, bins=10, color='skyblue')
plt.title("Histogram of Last Column")
plt.savefig("histogram.png")
plt.close()
```

```python
sns.scatterplot(x=x, y=y)
plt.title("Scatter Plot")
plt.xlabel(first_col)
plt.ylabel(last_col)
plt.savefig("scatter_plot.png")
plt.close()
```

```python
sns.boxplot(y=y)
plt.title("Box Plot of Last Column")
plt.savefig("box_plot.png")
plt.close()
```

```python
subset = df[[first_col, last_col]]
corr = subset.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.savefig("heatmap.png")
plt.close()
```
